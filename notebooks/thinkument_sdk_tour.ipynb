## Act 1: Connectivity & Environment Diagnostics

Thinkument operates on a **zero background sidecar** philosophy. You start your local studio daemon explicitly in a terminal (`thinkument dev` on `http://localhost:5052`), or supply a sovereign cloud API key (`THINKUMENT_API_KEY`). The SDK gracefully detects the active environment without hidden processes.

```python
import os
import sys
import requests
from thinkument_sdk import ThinkClient
from thinkument_sdk.types import (
    HealthzResponse,
    SessionLaunchResponse,
    SessionDispatchResponse,
    TapeEvent,
    JobStatusResponse,
)

# 1. Initialize client: auto-detects localhost:5052 or THINKUMENT_BASE_URL
client = ThinkClient()
daemon_online = False

try:
    health = client.healthz()
    daemon_online = True
    print("🟢 Thinkument Client Initialized & Connected to Live Studio")
    print(f"   Target Endpoint: {client.endpoint}")
    if '5052' in client.endpoint:
        tier_name = 'Tier 1: Free Local Studio ($0.00 In-Memory Loopback)'
    elif '8008' in client.endpoint or '8000' in client.endpoint:
        tier_name = 'Tier 2: Local Cloud Simulator (Django + PostgreSQL + Celery)'
    elif 'railway' in client.endpoint:
        tier_name = 'Tier 3: Railway Staging Cloud'
    else:
        tier_name = 'Tier 4: Sovereign Production Cloud (api.thinkument.com)'
    print(f"   Active Tier:     {tier_name}")
    print(f"   Service Status:  {health.status} (v{health.version})")
    print(f"   Authentication:  {'API Key Provisioned' if client.api_key else 'Anonymous / Offline'}")
except (requests.exceptions.ConnectionError, Exception) as e:
    daemon_online = False
    print("🟡 Thinkument Local Studio Daemon Offline (Explicit Terminal Lifecycle)")
    print(f"   Target Endpoint: {client.endpoint} (Unreachable)")
    print("   ℹ️  Zero Background Sidecars Invariant: Run 'thinkument dev' in another terminal")
    print("   ℹ️  Or export THINKUMENT_API_KEY='tk_live_...' to target Sovereign Cloud.")
    print("   ✨ Activating In-Cell Zero-Host Demonstration Simulation...")
```

## Act 2: Steve Jobsian Vibe Coding (`client.code.vibe`)

Traditional "vibe coding" produces thousands of lines of fragile imperative code. Thinkument transforms vibe coding into **Closed-World AST Synthesis**: an AI prompt synthesizes a mathematically sound, schema-strict $P \times SM \times C \times W$ manifest (`kind: think_program_v1`) via Operation `T6` in $<1\text{ms}$ locally.

```python
goal = "Editorial publishing workflow: author captures notes, AI drafts chapter, editor reviews"

if daemon_online:
    program = client.code.vibe(goal)
else:
    # Deterministic local AST simulation of Operation T6 (Closed-World AST Synthesis)
    program = {
        "kind": "think_program_v1",
        "run": {
            "program": {
                "info": {"title": "Editorial Publishing Flow", "description": goal},
                "flow": [
                    {"id": "capture", "widgets": ["author_notes"]},
                    {"id": "draft", "widgets": ["ai_chapter_drafter"]},
                    {"id": "review", "widgets": ["editorial_review"]},
                ],
                "widgets": {
                    "author_notes": {"widget": "note_collector", "mouth": {"text": "string"}, "belly": {}, "rear": {}},
                    "ai_chapter_drafter": {"widget": "llm_drafter", "mouth": {"outline": "string"}, "belly": {}, "rear": {}, "wires": {"outline": "author_notes.text"}},
                    "editorial_review": {"widget": "approval_gate", "mouth": {"draft": "string"}, "belly": {}, "rear": {}, "wires": {"draft": "ai_chapter_drafter.chapter"}},
                },
            }
        },
    }

print("✨ Program Synthesized Successfully!")
print(f"   Program Kind:    {program.get('kind')}")
print(f"   Title:           {program.get('run', {}).get('program', {}).get('info', {}).get('title')}")
print(f"   Declared Steps:  {len(program.get('run', {}).get('program', {}).get('flow', []))}")
```

## Act 3: Deep AST Inspection ($P \times SM \times C \times W$)

Every Thinkument program is composed of four canonical facets:
* **$P$ (Topology)**: Ordered steps, widget appearances, and dataflow wires (DAG).
* **$SM$ (Session Metadata)**: Role-modulated access matrix ($\Lambda$) and visibility rules.
* **$C$ (Context)**: Launch parameters and symbol resolution (`$symbols`).
* **$W$ (Widgets)**: Encapsulated components with Mouth ($M$), Belly ($B$), and Rear ($R$) ports.

> **The Law of Wire Typing Symmetry:** If widget $W_1$ emits `Promise[Draft]`, downstream widget $W_2$ receives unwrapped `Draft`. The kernel resolves promises out-of-band.

```python
prog_body = program.get("run", {}).get("program", {})
widgets = prog_body.get("widgets", {})
flow = prog_body.get("flow", [])

print("📋 Program Step Topology (P):")
for i, step in enumerate(flow, 1):
    step_id = step.get("id")
    step_widgets = step.get("widgets", [])
    print(f"   Step {i}: [{step_id}] -> mounted widgets: {step_widgets}")

print("\n🧩 Mounted Widgets & Wire Symmetry (W & Omega):")
for w_id, w_cfg in widgets.items():
    w_name = w_cfg.get("widget", "")
    wires = w_cfg.get("wires", {})
    print(f"   • {w_id} ({w_name})")
    print(f"     Incoming Wires: {wires or 'None (initial step ingestion)'}")
```

## Act 4: The "Micros on Reexploding" Encapsulation Proof

A foundational invariant of Thinkument language theory is **Component Encapsulation**:
1. **In Widget Authoring (`json_widget_v1`)**: When returning `reexplode`, the widget **must** specify `title` and `micros`. This tells the frontend womb what form inputs to render for the user's next turn.
2. **In Program Authoring (`think_program_v1`)**: The macro-program author is **strictly forbidden** from writing `micros`. Application flows must not leak or hardcode component presentation internals.

```python
def assert_no_key_recursively(obj, forbidden_key, path=""):
    if isinstance(obj, dict):
        for k, v in obj.items():
            curr = f"{path}.{k}" if path else k
            if k == forbidden_key:
                raise AssertionError(f"Encapsulation violation! Found '{k}' at {curr}")
            assert_no_key_recursively(v, forbidden_key, curr)
    elif isinstance(obj, list):
        for i, item in enumerate(obj):
            assert_no_key_recursively(item, forbidden_key, f"{path}[{i}]")

# Formally verify the Program AST adheres to the Encapsulation Theorem:
assert_no_key_recursively(program["run"]["program"], "micros", path="run.program")
print("✅ Seam Verified: Program AST contains zero 'micros' keys (Pure Macro-Composition).")
print("✅ Component Encapsulation: Widgets dynamically project micros during runtime reexplode.")
```

## Act 5: Invariant Program Evolution (`client.code.evolve`)

Thinkument enables developers to continuously mutate and evolve existing programs via natural language without risking non-deterministic code divergence. The AST mutation preserves all port typing, wire DAGs, and role mappings.

```python
instruction = "Add automated compliance and plagiarism screening step between draft and review"

if daemon_online:
    evolved = client.code.evolve(program, instruction=instruction)
else:
    import copy
    evolved = copy.deepcopy(program)
    prog = evolved["run"]["program"]
    prog["flow"].insert(2, {"id": "compliance", "widgets": ["compliance_screener"]})
    prog["widgets"]["compliance_screener"] = {
        "widget": "audit_screener",
        "mouth": {"text": "string"},
        "belly": {},
        "rear": {},
        "wires": {"text": "ai_chapter_drafter.chapter"},
    }
    prog["widgets"]["editorial_review"]["wires"]["draft"] = "compliance_screener.approved_text"

evolved_steps = evolved.get("run", {}).get("program", {}).get("flow", [])

print("🧬 Program Evolved Successfully!")
print(f"   Original Step Count: {len(flow)}")
print(f"   Evolved Step Count:  {len(evolved_steps)}")
for i, s in enumerate(evolved_steps, 1):
    print(f"   Step {i}: [{s.get('id')}] -> {s.get('widgets')}")
```

## Act 6: Instant 1-Liner Production Launch (`client.code.launch`)

With a single line of code, any synthesized or evolved program deploys into a **Live Collaborative Multi-User Session**. Roles and public view roles are automatically normalized.

```python
roles_cfg = {"author": {}, "editor": {}, "reviewer": {}}

if daemon_online:
    session = client.code.launch(evolved, mode="live", roles=roles_cfg)
else:
    import uuid
    sid = f"sess_{uuid.uuid4().hex[:8]}"
    session = SessionLaunchResponse(
        session_id=sid,
        url=f"{client.endpoint}/live/{sid}",
        status="active",
        roles=["author", "editor", "reviewer"],
        public_view_role="author",
    )

print("🚀 Live Collaborative Session Launched!")
print(f"   Session ID: {session.session_id}")
print(f"   Live URL:   {session.url or f'{client.endpoint}/live/{session.session_id}'}")
print(f"   Status:     {session.status}")
print(f"   Roles:      {session.roles or ['author', 'editor', 'reviewer']}")
```

## Act 7: Programmatic Poller-Dispatcher & Streaming (`client.session`)

Under the Thinkument architecture, client devices (browsers, mobile apps, or Python SDKs) are **zero-compute poller-dispatchers**. Dispatches execute in $<150\text{ms}$ with atomic Trident commits, while operational telemetry streams via epsilon channels.

```python
dispatch_payload = {"widget_id": "author_notes", "signal": "SUBMIT", "data": {"topic": "Chapter 1: The Genesis"}}

if daemon_online:
    dispatch_res = client.session.dispatch(session.session_id, dispatch_payload)
    events = client.session.events(session.session_id)
else:
    dispatch_res = SessionDispatchResponse(
        status="ok",
        session_id=session.session_id,
        applied_seq=1,
        latest_seq=1,
        data={"topic": "Chapter 1: The Genesis"},
    )
    events = [
        TapeEvent(channel="theta", payload={"step": "capture", "action": "SUBMIT"}, seq=1),
        TapeEvent(channel="epsilon", payload={"perf_ms": 12, "memory_kb": 1024}, seq=2),
    ]

print("⚡ Interactive Dispatch Applied in a Snap (<150ms SLA)!")
print(f"   Status:       {dispatch_res.status}")
print(f"   Applied Seq:  {dispatch_res.applied_seq}")
print(f"   Latest Seq:   {dispatch_res.latest_seq}")

print(f"\n📡 Telemetry Events Recorded ({len(events)}):")
for ev in events[:3]:
    print(f"   • [{ev.channel}] seq={ev.seq}: {ev.payload}")
```

## Act 8: Centralized Operations Relay & Universal Promise Probing (`client.ops`)

Custom cognitive operations execute through the **Universal Operations Relay** (`POST /api/v1/ops/execute/`).
Tasks are offloaded to the dedicated `QUEUE_ASYNCWORK` pool. External SDK clients probe status via the **Universal Promise-Probing Protocol** (`GET /api/v1/ops/jobs/<job_id>/`), maintaining complete symmetry with how widgets probe async promises internally.

```python
if daemon_online:
    handle = client.code.vibe_async("Parallel batch financial risk assessment flow")
    print(f"📦 Envelope Dispatched to Dedicated Pool (QUEUE_ASYNCWORK)!")
    print(f"   Job ID:   {handle.job_id}")
    print(f"   Poll URL: {handle.poll_url}")
    print("\n⏳ Probing Promise Status...")
    status_resp = handle.refresh()
    print(f"   Current Status: {status_resp.status}")
    final_job = handle.wait(timeout=10)
    print(f"✅ Promise Resolved! Status: {final_job.status}")
    print(f"   Synthesized Program Kind: {final_job.program.get('kind') if final_job.program else 'ok'}")
else:
    import uuid
    mock_job_id = f"job_{uuid.uuid4().hex[:8]}"
    print(f"📦 Envelope Dispatched to Dedicated Pool (QUEUE_ASYNCWORK)!")
    print(f"   Job ID:   {mock_job_id}")
    print(f"   Poll URL: /api/v1/ops/jobs/{mock_job_id}/")
    print("\n⏳ Probing Promise Status via Universal Promise Protocol...")
    print("   Current Status: running (seq=1)")
    print("✅ Promise Resolved! Status: success")
    print("   Synthesized Program Kind: think_program_v1")
```
